````python
import xml.etree.ElementTree as ET
import requests

year_url = f"https://home.treasury.gov/resource-center/data-chart-center/interest-rates/pages/xml?data=daily_treasury_yield_curve&field_tdr_date_value=2026"

response = requests.get(year_url)

print(response.status_code)
print(type(response.content))
````

````python
import time

url = (
    "https://home.treasury.gov/resource-center/"
    "data-chart-center/interest-rates/pages/xml"
)

params = {
    "data": "daily_treasury_yield_curve",
    "field_tdr_date_value": "2026",
}

start = time.perf_counter()

response = requests.get(
    url,
    params=params,
    timeout=30,
)

elapsed = time.perf_counter() - start

print(f"Status: {response.status_code}")
print(f"Response size: {len(response.content):,} bytes")
print(f"Elapsed: {elapsed:.2f} seconds")
print(f"Request URL: {response.url}")
````

````python
def print_xml_structure(element, depth=0):
    value = element.text.strip() if element.text and element.text.strip() else ""

    print("  " * depth + f"{element.tag}: {value}")

    for child in element:
        print_xml_structure(child, depth + 1)


root = ET.fromstring(response.content)
print_xml_structure(root)
````
