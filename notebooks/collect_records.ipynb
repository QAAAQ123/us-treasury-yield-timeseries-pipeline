```python
from extract import fetch_yield_curve_xml
import xml.etree.ElementTree as ET

xml_root = ET.fromstring(fetch_yield_curve_xml(2026, 9))

namespace = {
    "atom": "http://www.w3.org/2005/Atom",
    "m": "http://schemas.microsoft.com/ado/2007/08/dataservices/metadata",
}

properties_list = xml_root.findall(
    "atom:entry/atom:content/m:properties",
    namespace,
)

tags = {
    child.tag.split("}")[-1]: child.text
    for properties in properties_list
    for child in properties
}

print(tags["BC_5YEAR"])
```

```python
"""
records: list[dict[str, str]]
"""
import pprint
records = []
for property in properties_list:
    sub_dict = {
        value.tag.split("}")[-1]: value.text
        for value in property
    }
    records.append(sub_dict)

pprint.pprint(records[:2])
print(len(records))
```

```python
from datetime import date
cleaned_up_records: list[dict[str, str | None]] = []
for record in records:
    sub_dict = {
        key: value
        for key, value in record.items()
        if key == "NEW_DATE" or (key.startswith("BC_") and not key.endswith("DISPLAY"))
    }

    cleaned_up_records.append(sub_dict)

print(cleaned_up_records[:2])
print(cleaned_up_records[0].keys())
```
